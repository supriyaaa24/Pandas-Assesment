A CSV file contains commas inside quoted text fields, missing values, and dates in different formats. Explain how incorrect parsing can affect the resulting DataFrame and identify suitable read options or preprocessing steps.

## Given Dataset: students.csv

ID,Name,Date,Marks

Given Code

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
print(df.dtypes)
```

-> Comma in "Asha, R": It may be treated as a separator, causing extra columns unless the value is properly quoted.

-> Date column: Dates may be stored as text and need conversion to datetime.

-> Marks column: It may contain missing or non-numeric values.

## 3. Explanation of Issues
-> Quoted comma: "Asha, R" is read as one name.

-> Date column: Dates may have different formats, so convert them to datetime.

-> Marks column: Blank and NA values are treated as missing, so Marks becomes a float column.

-> NA value: Check whether NA means missing or something else before replacing it.

### Corrected code

```python
import pandas as pd

# Preserve NA temporarily to investigate its meaning
df = pd.read_csv("students.csv", keep_default_na=False)

# Convert blank marks to missing values
df['Marks'] = df['Marks'].replace('', pd.NA)

# Convert marks into numeric values
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')

# Convert dates into a consistent datetime format
df['Date'] = pd.to_datetime(
    df['Date'], format='mixed', dayfirst=True, errors='coerce'
)

print(df)
print(df.dtypes)
print(df.isnull().sum())
```

### Explanation of Corrected Code
-> keep_default_na=False → keeps "NA" as text for investigation.

-> replace('', pd.NA) → converts blanks to missing values.

-> pd.to_numeric(errors='coerce') → converts marks to numbers; invalid values become NaN.

-> pd.to_datetime() → converts dates to datetime format.

-> dayfirst=True → reads 01/09/2026 as 1 September 2026.

-> errors='coerce' → invalid dates become NaT.

-> isnull().sum() → counts missing values.

-> format='mixed' → handles different date formats; ambiguous dates should be verified.



### Importance of Investigating NA
-> NA may mean missing, absent, or another value.

-> Assuming it is missing can lead to wrong results.

-> First verify its meaning, then handle it appropriately.

###  Given Code

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
```

#### Median: 
The median is the middle value after sorting the salaries. It is 32,500, which is much lower than the mean.

#### Effect of the extreme value:
 The mean is highly affected by the extreme salary, the median is relatively resistant to outliers.

#### Representative statistic: 
The median may better represent a typical employee's salary because it is less affected by the unusually high salary.

### Query-1: Student Assessment Data Cleaning
#### 1. Question
Using Pandas, perform the following cleaning operations in sequence:

-> Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().

-> Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

-> Identify and remove exact duplicate student assessment records.

-> Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.

-> Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

-> Identify marks outside the valid range of 0–100 and decide how they should be handled.

-> Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

## 2. Answer
##### Step 1: Import Libraries and Load the Dataset

```python
import pandas as pd

# Load the CSV file while preserving NA and blank values
df = pd.read_csv(
    "student_assessment_dirty.csv",
    keep_default_na=False
)

# Inspect the dataset
print(df.head())
print(df.info())
print(df.isnull().sum())
```

### Step 2: Standardize Department Names

```python
import pandas as pd

df = pd.read_csv("student_assessment_dirty.csv", sep="\t")

df['Department'] = df['Department'].str.strip().str.upper()

df['Department'] = df['Department'].replace({
    'I.T.': 'IT'
})

print(df['Department'].unique())
```

## Step 3: Identify and Remove Duplicates

```python
# Count exact duplicate records
duplicates = df.duplicated().sum()

# Remove exact duplicates
df = df.drop_duplicates()

print("Records removed:", duplicates)
```

### Step 4: Convert Marks into Numeric Format

```python
# Clean spaces and standardize text
df['Marks'] = df['Marks'].astype('string').str.strip()

# Preserve the original NA and blank values for classification
marks_text = df['Marks'].str.upper()

# Identify absent students separately
df['Attendance_Status'] = marks_text.map({
    'ABSENT': 'Absent'
}).astype('string')

# Classify original missing-value representations
df['Marks_Status'] = 'Available'

df.loc[marks_text.eq('NA'), 'Marks_Status'] = 'NA value'
df.loc[df['Marks'].eq(''), 'Marks_Status'] = 'Blank'
df.loc[marks_text.eq('ABSENT'), 'Marks_Status'] = 'Absent'

# Remove the word 'marks' from numeric entries
df['Marks'] = df['Marks'].str.replace(
    r'\s*marks$', '', regex=True, case=False
)

# Convert marks into numeric values
df['Marks'] = pd.to_numeric(
    df['Marks'].replace({
        'NA': pd.NA,
        '': pd.NA,
        'Absent': pd.NA,
        'ABSENT': pd.NA
    }),
    errors='coerce'
)

print(df[['Student_Name', 'Marks', 'Marks_Status',
          'Attendance_Status']])
```

### Step 5: Investigate Missing Marks

```python
# Count absent students
absent = df['Attendance_Status'].eq('Absent').sum()

# Count missing marks
missing = df['Marks'].isnull().sum()

# Identify students whose marks have not been entered
not_entered = df[
    df['Marks'].isnull() &
    df['Attendance_Status'].isna()
]

# Count original NA and blank entries
na_count = df['Marks_Status'].eq('NA value').sum()
blank_count = df['Marks_Status'].eq('Blank').sum()

print("Absent students:", absent)
print("Missing marks:", missing)
print("Marks not entered:", len(not_entered))
print("Original NA values:", na_count)
print("Original blank values:", blank_count)
```

### Step 6: Identify Invalid Marks

```python
# Identify marks outside the valid range
invalid = df[
    df['Marks'].notnull() &
    ~df['Marks'].between(0, 100)
]

print("Invalid marks:")
print(invalid[['Student_Name', 'Marks']])
```

### Step 7: Display the Cleaned DataFrame

```python
# Display the cleaned dataset
print(df)
```

### 4. Explanation

#### Dataset inspection: 
head() displays the first five rows, info() provides information about columns and data types, and isnull().sum() counts values already recognized as missing. Since keep_default_na=False is used, the original NA and blank values are preserved during loading.

#### Department standardization:
 str.strip(), str.upper() and replace() standardize inconsistent department names, allowing them to be grouped consistently.

#### Duplicate removal: 
duplicated() identifies exact duplicate rows, and drop_duplicates() removes them while retaining the first occurrence.

#### Numeric conversion: 
str.replace() removes the word marks from entries such as 88 marks. pd.to_numeric(errors='coerce') converts numeric strings into numbers and unrecognized values into NaN.

#### Investigating NA and blank values:
 keep_default_na=False prevents Pandas from automatically converting NA and blank entries into missing values during loading. They are classified separately before conversion, preserving information about their original representations.

#### Missing marks: 
Absent is identified separately from students whose marks have not been entered. Both are represented as missing numeric marks, but their status is retained in separate columns.

#### Risk of deleting missing values:
 Deleting missing records without investigating them may remove valid student records and introduce bias into the analysis.

#### Invalid marks: 
Marks outside the valid range of 0–100 are identified for investigation. The value 105 is retained and flagged. If it is a data-entry error, it should be corrected using the verified value. If it is confirmed to be invalid, it should be excluded from valid-mark calculations. A score of 0 is valid.
